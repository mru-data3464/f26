```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

from sklearn.model_selection import cross_val_score
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import make_column_transformer
from sklearn.pipeline import make_pipeline
```

```python
df = pd.read_csv("titanic_train.csv")
df.head()
```

```python
df.info()
```

```python
# Select just the features that don't have nulls (we'll revisit these at a later date)
num_feats = ["Pclass", "SibSp", "Parch", "Fare"]
cat_feats = ["Sex", "Embarked"]
y = df["Survived"]
```

```python
# Basic EDA
pd.plotting.scatter_matrix(df[num_feats], c=y)
plt.show()

# categorical data might need some fancier plotting
# seaborn's catplot automates a lot of this
for cat in cat_feats:
    counts = df.groupby(cat)["Survived"].value_counts()
    plt.figure()
    x = np.arange(0, len(counts) / 2) * 1.25
    plt.bar(x-.25, counts[:,0], width=0.5, label="Did not survive")
    plt.bar(x+.25, counts[:,1], width=0.5, label="Survived")
    plt.xticks(x, labels=np.unique([i[0] for i in counts.index]))
    plt.xlabel(counts.index.names[0])
    plt.ylabel("Number of passengers")
    plt.legend()
```
